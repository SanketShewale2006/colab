```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("/content/messy_customer_data.csv")
df
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df['Name']=df['Name'].fillna(df['Name'].mode()[0])
```

```python
df['Age']=df['Age'].fillna(df['Age'].mean())
df['Age']=df['Age'].astype(int)
```

```python
df['Gender']=df['Gender'].fillna(df['Gender'].mode()[0])
```

```python
df['City']=df['City'].fillna(df['City'].mode()[0])
df.head(2)
```

```python
df['Purchase_Amount'] = pd.to_numeric(df['Purchase_Amount'], errors='coerce')
df['Purchase_Amount'] = df['Purchase_Amount'].fillna(100)
df['Purchase_Amount'] = df['Purchase_Amount'].astype(int)
```

```python
df['Name']=df['Name'].str.upper()
df['City']=df['City'].str.lower()
df['Gender']=df['Gender'].str.upper()
```

```python
gender_replacements = {
    'M': 'MALE',
    'F': 'FEMALE'
}
df['Gender'] = df['Gender'].replace(gender_replacements)
display(df.head())
```

```python
df = df.sort_values(by='Customer_ID',ascending=True)
df = df.reset_index(drop=True)
```

```python
df['Age'] = df['Age'].abs()
```

```python
df
```

```python
def categorize_age(age):
    if age <= 12:
        return 'Child'
    elif 12 < age < 18:
        return 'Teenager'
    else:
        return 'Adult'

df['Catogery'] = df['Age'].apply(categorize_age)
df
```

```python
df['Purchase_Amount'] = df['Purchase_Amount'].abs()
```

```python
df.describe()
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
X = df[['Age']]
y = df['Purchase_Amount']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=20, random_state=42)
model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)*100

print(f'Mean Squared Error: {mse:.2f}')
print(f'R-squared: {r2:.2f}')

```

```python
new_data = float(input("enter the age of purchase person : "))

new_data_reshaped = np.array(new_data).reshape(-1, 1)
predicted_purchase_amount = model.predict(new_data_reshaped)
print(f"Predicted purchase amount for age {int(new_data)}: {predicted_purchase_amount[0]:.2f}")
```

```python
# Optionally, visualize the results
plt.scatter(X_test, y_test, color='blue', label='Actual Purchase Amount')
plt.plot(X_test, y_pred, color='red', linewidth=2, label='Predicted Purchase Amount')
plt.xlabel('Age')
plt.ylabel('Purchase Amount')
plt.title('Age vs. Purchase Amount (Linear Regression)')
plt.legend()
plt.show()
```
